```python
# === v4 IPYWIDGETS QUIZ SETUP ===
# Run this cell FIRST (Shift+Enter). All code here is HIDDEN from students.
import base64, json, time
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# Inject CSS to hide code cells and style the quiz beautifully
display(HTML("""
<style>
/* Hide all code input cells in JupyterLab */
.jp-Cell .jp-Cell-inputWrapper { display: none !important; }
.jp-Notebook .jp-Cell code { display: none !important; }

/* Quiz card styling — neon arcade theme */
.quiz-card {
  background: linear-gradient(135deg, #14142B 0%, #1F1F3D 100%);
  border: 1px solid #00E5FF;
  border-radius: 12px;
  padding: 20px;
  margin: 12px 0;
  box-shadow: 0 0 20px rgba(0, 229, 255, 0.15), 0 4px 16px rgba(0,0,0,0.4);
  font-family: 'Inter', sans-serif;
  color: #F4F4F8;
}
.quiz-card-incorrect {
  border-color: #FF2E9A;
  box-shadow: 0 0 20px rgba(255, 46, 154, 0.2), 0 4px 16px rgba(0,0,0,0.4);
}
.quiz-card-correct {
  border-color: #00E676;
  box-shadow: 0 0 20px rgba(0, 230, 118, 0.2), 0 4px 16px rgba(0,0,0,0.4);
}
.quiz-question-text {
  font-size: 14px;
  line-height: 1.6;
  color: #F4F4F8;
  margin-bottom: 12px;
}
.quiz-hint {
  font-size: 12px;
  color: #00E5FF;
  font-style: italic;
  margin-top: 8px;
}
.quiz-feedback-correct {
  background: rgba(0, 230, 118, 0.15);
  border-left: 3px solid #00E676;
  color: #00E676;
  padding: 8px 12px;
  margin-top: 8px;
  border-radius: 4px;
  font-weight: 600;
}
.quiz-feedback-incorrect {
  background: rgba(255, 46, 154, 0.15);
  border-left: 3px solid #FF2E9A;
  color: #FF2E9A;
  padding: 8px 12px;
  margin-top: 8px;
  border-radius: 4px;
  font-weight: 600;
}
.quiz-progress-bar {
  position: sticky;
  top: 0;
  z-index: 100;
  background: rgba(11, 11, 31, 0.95);
  padding: 12px 16px;
  border-radius: 8px;
  margin-bottom: 16px;
  border: 1px solid #2A2A4E;
}
.quiz-tab-content {
  padding: 16px;
}
.quiz-submit-button {
  background: linear-gradient(135deg, #FF2E9A 0%, #FFD60A 100%) !important;
  color: #0B0B1F !important;
  font-weight: 700 !important;
  font-size: 16px !important;
  padding: 12px 32px !important;
  border-radius: 8px !important;
  border: none !important;
  box-shadow: 0 0 24px rgba(255, 46, 154, 0.5) !important;
  cursor: pointer !important;
}
.quiz-submit-button:hover {
  box-shadow: 0 0 32px rgba(255, 46, 154, 0.7) !important;
  transform: translateY(-2px);
}
.quiz-analytics {
  background: linear-gradient(135deg, #14142B 0%, #1F1F3D 100%);
  border: 1px solid #00E5FF;
  border-radius: 12px;
  padding: 24px;
  margin: 16px 0;
  color: #F4F4F8;
  font-family: 'JetBrains Mono', monospace;
  font-size: 13px;
  line-height: 1.6;
  box-shadow: 0 0 24px rgba(0, 229, 255, 0.15);
  white-space: pre-wrap;
}
.quiz-analytics h2 {
  color: #FF2E9A;
  font-family: 'Press Start 2P', monospace;
  font-size: 18px;
  margin-top: 16px;
  margin-bottom: 8px;
  text-shadow: 0 0 12px rgba(255, 46, 154, 0.6);
}
.quiz-analytics h3 {
  color: #00E5FF;
  font-family: 'Press Start 2P', monospace;
  font-size: 14px;
  margin-top: 12px;
  margin-bottom: 6px;
}
.quiz-analytics .correct { color: #00E676; }
.quiz-analytics .incorrect { color: #FF2E9A; }
.quiz-analytics .highlight { color: #FFD60A; font-weight: 700; }


/* === v5 FIX 4: COLOR CONTRAST FIX === */
/* Force light labels on ALL widget types — fixes Part B onwards contrast */

/* Radio button labels (Part B, D, E, F, H) */
.widget-radio-box label,
.widget-radio label,
.widget-RadioButtons label {
    color: #F4F4F8 !important;
    font-size: 13px !important;
    line-height: 1.5 !important;
}

/* Dropdown labels and descriptions (Part C, G, name fields) */
.widget-dropdown label,
.widget-Dropdown label,
.widget-dropdown-box label {
    color: #F4F4F8 !important;
    font-size: 13px !important;
}

.widget-dropdown select {
    background-color: #1F1F3D !important;
    color: #F4F4F8 !important;
    border: 1px solid #00E5FF !important;
    font-size: 13px !important;
}

/* Text input labels (Part A, name fields) */
.widget-text label,
.widget-Text label,
.widget-FloatText label,
.widget-text-box label {
    color: #F4F4F8 !important;
    font-size: 13px !important;
}

.widget-text input,
.widget-FloatText input,
.widget-Text input {
    background-color: #1F1F3D !important;
    color: #F4F4F8 !important;
    border: 1px solid #00E5FF !important;
    font-size: 14px !important;
    padding: 4px 8px !important;
}

/* Generic widget labels */
.widget-label,
.widget-label-basic {
    color: #F4F4F8 !important;
    font-size: 13px !important;
}

/* Radio button circles — ensure visible */
.widget-radio input[type=radio] {
    accent-color: #FF2E9A !important;
    transform: scale(1.2);
    margin-right: 8px !important;
}

/* Button styling — wider, more prominent */
.widget-button {
    min-width: 160px !important;
    min-height: 40px !important;
    font-size: 13px !important;
    font-weight: 600 !important;
    border-radius: 6px !important;
}

/* Submit button — extra prominent */
.quiz-submit-button,
.widget-button.quiz-submit-button {
    min-width: 240px !important;
    min-height: 56px !important;
    font-size: 16px !important;
    padding: 12px 32px !important;
    width: auto !important;
}

/* Progress bar — better visibility */
.widget-progress-bar .progress {
    background-color: #1F1F3D !important;
}

.widget-progress-bar .progress-bar {
    background-color: #00E5FF !important;
}

/* Accordion (hints) — better contrast */
.widget-accordion .accordion-button {
    color: #00E5FF !important;
    background-color: #14142B !important;
}

.widget-accordion .accordion-item {
    background-color: #1F1F3D !important;
    border: 1px solid #2A2A4E !important;
}

/* Tab styling — better contrast */
.widget-tab .widget-tab-contents {
    background-color: #0B0B1F !important;
}

.widget-tab .widget-tab-bar {
    background-color: #14142B !important;
}

.widget-tab button {
    color: #9494B8 !important;
    background-color: #14142B !important;
    border: 1px solid #2A2A4E !important;
}

.widget-tab button.widget-tab-tab-selected {
    color: #FF2E9A !important;
    background-color: #1F1F3D !important;
    border-bottom: 2px solid #FF2E9A !important;
}

/* VBox/HBox — ensure children inherit proper styling */
.widget-vbox,
.widget-hbox {
    color: #F4F4F8 !important;
}

/* Name input field — special styling */
.name-input-card {
    background: linear-gradient(135deg, #14142B 0%, #1F1F3D 100%);
    border: 1px solid #FFD60A;
    border-radius: 12px;
    padding: 20px;
    margin: 12px 0;
    box-shadow: 0 0 20px rgba(255, 214, 10, 0.15), 0 4px 16px rgba(0,0,0,0.4);
}


/* === v6 FIX 4: RADIO BUTTON SPACING (fixes overlap) === */
/* Ensure radio button options don't overlap — each on its own line with proper spacing */
.widget-radio-box {
    row-gap: 10px !important;
    padding: 8px 0 !important;
    display: flex !important;
    flex-direction: column !important;
}

.widget-radio-box .widget-radio {
    margin-bottom: 8px !important;
    padding: 4px 0 !important;
}

.widget-radio-box label,
.widget-radio label {
    line-height: 1.6 !important;
    white-space: normal !important;
    display: block !important;
    margin-bottom: 6px !important;
    color: #F4F4F8 !important;
    font-size: 13px !important;
    word-wrap: break-word !important;
    overflow-wrap: break-word !important;
}

.widget-radio input[type=radio] {
    margin-right: 8px !important;
    margin-bottom: 4px !important;
    vertical-align: top !important;
}

/* Also fix dropdown option text wrapping */
.widget-dropdown-box {
    padding: 4px 0 !important;
}

.widget-dropdown select option {
    color: #F4F4F8 !important;
    background-color: #1F1F3D !important;
    padding: 4px !important;
}


/* === v7 FIX 1 & 2: RADIO BUTTON TOP-ALIGNMENT === */
/* Fix: radio circle appears in middle of wrapped text. Align to TOP instead. */
.widget-radio {
    align-items: flex-start !important;
    -webkit-align-items: flex-start !important;
    display: flex !important;
    flex-direction: row !important;
}

.widget-radio input[type=radio] {
    margin-top: 3px !important;
    margin-right: 8px !important;
    flex-shrink: 0 !important;
    -webkit-flex-shrink: 0 !important;
}

.widget-radio label {
    flex-grow: 1 !important;
    -webkit-flex-grow: 1 !important;
    margin-top: 0 !important;
}
</style>
"""))

# === ENCODED ANSWER KEYS (base64) ===
_NUM_ANSWERS = {'q1': 'NS4w', 'q2': 'NDUwMC4w', 'q3': 'NDAwMC4w', 'q4': 'MjAuMA==', 'q5': 'OS40Mg==', 'q6': 'NTAuMA==', 'q7': 'NDkuMA==', 'q8': 'NS4w'}
_MC_ANSWERS = {'cp1': 'Qg==', 'cp2': 'Qg==', 'cp3': 'QQ==', 'mc1': 'Qg==', 'mc2': 'Qg==', 'es1': 'Qg==', 'es2': 'QQ==', 'ui1': 'Qg==', 'ui2': 'Qw==', 'ei1': 'Qg=='}
_MATCH_ANSWERS = {'mr1': 'eyJtcjFhIjogIkEiLCAibXIxYiI6ICJCIiwgIm1yMWMiOiAiQyIsICJtcjFkIjogIkQifQ=='}
_ORDER_ANSWERS = {'os1': 'WyJCIiwgIkQiLCAiQyIsICJBIiwgIkUiXQ=='}

# === v3 TIMING + ATTEMPT TRACKING (preserved) ===
_QUIZ_START_TIME = time.time()
_QUIZ_SUBMITTED = False
_QUIZ_END_TIME = None
_ATTEMPT_LOG = []
_ATTEMPT_COUNT = {}
_LAST_ATTEMPT_TIME = {}
_SCORES = {}  # Track final correctness per question

# === v4 NEW: Widget state tracking ===
_ANSWERED_QUESTIONS = set()  # qids that have been answered at least once
_WIDGET_REGISTRY = {}  # qid → {widget, check_button, feedback_html, hint_accordion}
_TEACHER_VERSION = False

# Topic-specific feedback data (preserved from v3)
_TOPIC_FEEDBACK = {"mnemonics": [["F = ma", "'Force = mass times Acceleration' \u2014 Newton's 2nd"], ["1st: Stay", "'1st Law: Stay' \u2014 objects at rest stay at rest"], ["2nd: Push", "'2nd Law: Push' \u2014 F = ma"], ["3rd: React", "'3rd Law: React' \u2014 equal and opposite"], ["Weight = mg", "'Weight = mg' (NOT mass alone)"]], "type_advice": {"A": {"title": "Numerical Calculations (Part A)", "common_errors": ["Confusing mass (kg) with weight (N) \u2014 weight = m \u00d7 g", "Forgetting to square or take square root in kinematic formulas", "Using F = mv (momentum) when you should use F = ma (force)", "Sign errors in 3rd Law problems (forces are opposite)"], "step_by_step_example": "Example: 'm = 2 kg, F = 10 N. Find a.'\n  Step 1: Goal \u2014 acceleration.\n  Step 2: Use Newton's 2nd Law (F = ma).\n  Step 3: Known: m=2 kg, F=10 N. Unknown: a (m/s\u00b2).\n  Step 4: a = F/m = 10/2 = 5 m/s\u00b2.", "tips": ["F = ma \u2192 a = F/m (DIVIDE). Mass \u00d7 g = weight (N, not kg)"]}, "B": {"title": "Computational Procedure (Part B)", "common_errors": ["Confusing Newton's 1st, 2nd, 3rd laws"], "step_by_step_example": "1st: inertia (no force = no change). 2nd: F = ma (force causes acceleration). 3rd: action-reaction pairs.", "tips": ["1st = no force. 2nd = how much force. 3rd = reactions"]}, "C": {"title": "Match-the-Reasoning (Part C)", "common_errors": ["Matching 1st Law problem to 2nd Law formula"], "step_by_step_example": "Find acceleration \u2192 2nd Law (F=ma). Find reaction force \u2192 3rd Law (F_action = -F_reaction). Find weight \u2192 W = mg.", "tips": ["Match the law to the question being asked"]}, "D": {"title": "Conceptual Multiple Choice (Part D)", "common_errors": ["Thinking 3rd Law forces cancel (they act on DIFFERENT objects!)", "Confusing mass with weight", "Thinking 'no motion = no force' (could be balanced forces)"], "step_by_step_example": "3rd Law pairs act on different objects \u2014 they don't cancel. You push wall, wall pushes you.", "tips": ["3rd Law: same magnitude, opposite direction, DIFFERENT objects"]}, "E": {"title": "Equation Selection (Part E)", "common_errors": ["Using F = mv (momentum) instead of F = ma"], "step_by_step_example": "Force: F = ma. Weight: W = mg. Momentum: p = mv. Different quantities!", "tips": ["Memorize: F = ma (force), W = mg (weight), p = mv (momentum)"]}, "F": {"title": "Unit Identification (Part F)", "common_errors": ["Confusing kg (mass) with N (force)", "Forgetting that 1 N = 1 kg\u00b7m/s\u00b2"], "step_by_step_example": "Force in N. Mass in kg. Acceleration in m/s\u00b2. Weight in N (force!).", "tips": ["1 N = 1 kg \u00d7 1 m/s\u00b2. Mass is in kg; weight is in N"]}, "G": {"title": "Order the Steps (Part G)", "common_errors": ["Computing force before finding acceleration"], "step_by_step_example": "(1) Knowns, (2) state unknown, (3) find a from kinematics, (4) compute F = ma, (5) verify.", "tips": ["Often need kinematics first, THEN F = ma"]}, "H": {"title": "Error Identification (Part H)", "common_errors": ["Multiply-vs-divide in a = F/m (should DIVIDE, not multiply)"], "step_by_step_example": "Student wrote a = F \u00d7 m. Should be a = F/m. Inverted the operation.", "tips": ["a = F/m. Force ON TOP, mass on BOTTOM"]}}, "advanced_problems": [{"title": "Rocket Thrust: Newton's 3rd Law in Action", "problem": "A rocket expels exhaust gas at 3000 m/s (relative to rocket). The rocket mass is 100,000 kg initially, and it burns 1000 kg of fuel per second. (a) Find the thrust force (use F_thrust = \u1e41 \u00d7 v_exhaust, where \u1e41 is mass flow rate). (b) Find initial acceleration. (c) As fuel burns, mass decreases \u2014 does acceleration increase or decrease?", "hint": "F = \u1e41 \u00d7 v_e = 1000 \u00d7 3000 = 3,000,000 N. a = F/m = 3e6/1e5 = 30 m/s\u00b2."}, {"title": "Elevator Forces: Apparent Weight", "problem": "A 70 kg person stands in an elevator. (a) Find 'apparent weight' (normal force from floor) when elevator accelerates UP at 2 m/s\u00b2. (b) When accelerating DOWN at 2 m/s\u00b2. (c) When in free-fall (cable breaks). (d) When moving at constant velocity (up or down).", "hint": "N = m \u00d7 (g + a) for up-acceleration, N = m \u00d7 (g - a) for down, N = 0 in free-fall, N = mg at constant velocity."}, {"title": "William as Unbalanced Force", "problem": "Model William's village as an object with mass m = 1000 (metaphorical 'inertia' of tradition). William applies F = 50 N (effort) for t = 5 years. (a) Find the village's 'acceleration' a = F/m. (b) Find final 'velocity' v = a \u00d7 t. (c) If William doubles his effort (F = 100 N), how much faster does the village change?", "hint": "a = F/m. v = at. Linear scaling: doubling F doubles a and v."}]}

# Type prefix map for analytics
_TYPE_MAP = {"A": "Numerical", "B": "Computational Procedure", "C": "Match-the-Reasoning", "D": "Conceptual MC", "E": "Equation Selection", "F": "Unit Identification", "G": "Order the Steps", "H": "Error Identification"}

# === GRADING FUNCTIONS (preserved from v3, with widget integration) ===

def _log_attempt(qid, qtype, student_answer, was_correct, partial_score=None, partial_total=None):
    """Log an attempt to the attempt log."""
    ts = time.time()
    gap = ts - LAST_ATTEMPT_TIME.get(qid, ts)
    LAST_ATTEMPT_TIME[qid] = ts
    attempt_num = _ATTEMPT_COUNT.get(qid, 0) + 1
    _ATTEMPT_COUNT[qid] = attempt_num
    entry = {
        "qid": qid, "qtype": qtype, "timestamp": ts,
        "student_answer": str(student_answer), "was_correct": was_correct,
        "attempt_num": attempt_num, "gap_since_last": gap
    }
    if partial_score is not None:
        entry["partial_score"] = partial_score
        entry["partial_total"] = partial_total
    _ATTEMPT_LOG.append(entry)
    _ANSWERED_QUESTIONS.add(qid)
    if was_correct:
        _SCORES[qid] = True
    elif qid not in _SCORES:
        _SCORES[qid] = False
    return attempt_num

def grade_numerical(qid, student_answer, tolerance=0.02):
    """Grade numerical answer."""
    if _QUIZ_SUBMITTED:
        return False, "Quiz submitted. Answers frozen."
    correct = float(base64.b64decode(_NUM_ANSWERS[qid]).decode())
    try:
        s = float(student_answer)
    except (TypeError, ValueError):
        _log_attempt(qid, "A", student_answer, False)
        return False, "⚠ Please enter a numerical answer."
    if abs(s - correct) / max(abs(correct), 1e-9) < tolerance:
        _log_attempt(qid, "A", s, True)
        return True, "✓ CORRECT! Great work."
    else:
        _log_attempt(qid, "A", s, False)
        if s > correct:
            return False, "✗ Not quite. Your answer is too HIGH. Try again."
        else:
            return False, "✗ Not quite. Your answer is too LOW. Try again."

def grade_mc(qid, student_choice):
    """Grade multiple choice."""
    if _QUIZ_SUBMITTED:
        return False, "Quiz submitted. Answers frozen."
    correct = base64.b64decode(_MC_ANSWERS[qid]).decode()
    qtype_map = {"q": "A", "cp": "B", "mr": "C", "mc": "D", "es": "E", "ui": "F", "os": "G", "ei": "H"}
    qtype_prefix = qid[:2] if qid[:2] in ["cp", "mr", "mc", "es", "ui", "os", "ei"] else qid[0]
    qtype = qtype_map.get(qtype_prefix, "?")
    if str(student_choice).strip().upper() == correct:
        _log_attempt(qid, qtype, student_choice, True)
        return True, "✓ CORRECT! Great work."
    else:
        _log_attempt(qid, qtype, student_choice, False)
        return False, "✗ Not quite. Try a different option."

def grade_match(qid, student_mapping):
    """Grade match-the-reasoning."""
    if _QUIZ_SUBMITTED:
        return 0, 0, "Quiz submitted."
    correct_mapping = json.loads(base64.b64decode(_MATCH_ANSWERS[qid]).decode())
    n_correct = 0
    n_total = len(correct_mapping)
    for sub_q, correct_letter in correct_mapping.items():
        student_letter = str(student_mapping.get(sub_q, "?")).strip().upper()
        if student_letter == correct_letter:
            n_correct += 1
    _log_attempt(qid, "C", student_mapping, (n_correct == n_total), n_correct, n_total)
    if n_correct == n_total:
        return n_correct, n_total, "✓ ALL CORRECT! " + str(n_correct) + "/" + str(n_total)
    else:
        return n_correct, n_total, "✗ " + str(n_correct) + "/" + str(n_total) + " correct. Try again."

def grade_order(qid, student_order):
    """Grade order-the-steps."""
    if _QUIZ_SUBMITTED:
        return 0, 0, "Quiz submitted."
    correct_order = json.loads(base64.b64decode(_ORDER_ANSWERS[qid]).decode())
    n_correct = 0
    n_total = len(correct_order)
    for s, c in zip(student_order, correct_order):
        if str(s).strip() == str(c).strip():
            n_correct += 1
    _log_attempt(qid, "G", student_order, (n_correct == n_total), n_correct, n_total)
    if n_correct == n_total:
        return n_correct, n_total, "✓ PERFECT ORDER!"
    else:
        return n_correct, n_total, "✗ " + str(n_correct) + "/" + str(n_total) + " in place. Try again."

# === v4 NEW: Widget rendering functions ===

def _create_feedback_html():
    """Create an empty feedback HTML widget."""
    return widgets.HTML(value="", placeholder="Feedback will appear here...")

def _create_hint_accordion(hint_text):
    """Create a collapsible hint accordion."""
    hint_html = widgets.HTML(value='<div class="quiz-hint">💡 ' + hint_text + '</div>')
    accordion = widgets.Accordion(children=[hint_html])
    accordion.set_title(0, '💡 Show Hint (click to expand)')
    accordion.selected_index = None  # Collapsed by default
    return accordion

def _update_feedback(feedback_widget, is_correct, message, card_widget=None):
    """Update the feedback HTML widget with styled message."""
    if is_correct:
        feedback_widget.value = '<div class="quiz-feedback-correct">' + message + '</div>'
        if card_widget:
            card_widget.add_class('quiz-card-correct')
            card_widget.remove_class('quiz-card-incorrect')
    else:
        feedback_widget.value = '<div class="quiz-feedback-incorrect">' + message + '</div>'
        if card_widget:
            card_widget.add_class('quiz-card-incorrect')
            card_widget.remove_class('quiz-card-correct')
    _update_progress_bar()



# === v5 FIX 2: STUDENT NAME FIELDS (Pinyin + English) ===
_STUDENT_NAME_PINYIN = ""
_STUDENT_NAME_ENGLISH = ""

def render_name_input_card():
    'Render the student name input card at the top of the dashboard.'
    pinyin_input = widgets.Text(value="", placeholder='e.g., Zhang San', description='Name (Pinyin):')
    english_input = widgets.Text(value="", placeholder='e.g., John Zhang', description='Name (English):')
    pinyin_input.style.description_width = '150px'
    english_input.style.description_width = '150px'

    # Increase input width
    pinyin_input.layout = widgets.Layout(width='400px')
    english_input.layout = widgets.Layout(width='400px')

    def on_pinyin_change(change):
        global _STUDENT_NAME_PINYIN
        _STUDENT_NAME_PINYIN = change['new']

    def on_english_change(change):
        global _STUDENT_NAME_ENGLISH
        _STUDENT_NAME_ENGLISH = change['new']

    pinyin_input.observe(on_pinyin_change, names='value')
    english_input.observe(on_english_change, names='value')

    name_html = widgets.HTML(value='<div style="color: #FFD60A; font-size: 16px; font-weight: 700; margin-bottom: 12px;">👤 Student Information</div><div style="color: #9494B8; font-size: 12px; margin-bottom: 16px;">Please enter your name before starting the quiz.</div>')

    card = widgets.VBox([name_html, pinyin_input, english_input])
    card.add_class('name-input-card')
    return card


def _update_progress_bar():
    """Update the global progress bar and score tracker."""
    if '_PROGRESS_BAR' in globals() and '_SCORE_TRACKER' in globals():
        answered = len(_ANSWERED_QUESTIONS)
        total = len(_NUM_ANSWERS) + len(_MC_ANSWERS) + sum(len(json.loads(base64.b64decode(v).decode())) for v in _MATCH_ANSWERS.values()) + sum(len(json.loads(base64.b64decode(v).decode())) for v in _ORDER_ANSWERS.values())
        _PROGRESS_BAR.value = answered
        _PROGRESS_BAR.max = total
        correct_count = sum(1 for v in _SCORES.values() if v)
        _SCORE_TRACKER.value = '<div style="color: #00E5FF; font-family: monospace; font-size: 14px;">📊 Answered: ' + str(answered) + '/' + str(total) + ' | ✓ Correct: ' + str(correct_count) + '</div>'

def render_numerical_question(qid, question_text, units, hint_text=""):
    """Render a numerical question as a widget card."""
    question_html = widgets.HTML(value='<div class="quiz-question-text">' + question_text + '</div>')
    input_widget = widgets.FloatText(value=0, description='Answer:', placeholder='Enter number')
    units_label = widgets.HTML(value='<div style="color: #FFD60A; margin-left: 8px; font-family: monospace;">' + units + '</div>')
    check_button = widgets.Button(description='✓ Check Answer', button_style='info', layout=widgets.Layout(min_width='180px', height='40px'))
    feedback_html = _create_feedback_html()

    input_row = widgets.HBox([input_widget, units_label, check_button])

    card_children = [question_html, input_row, feedback_html]
    if hint_text:
        card_children.append(_create_hint_accordion(hint_text))

    card = widgets.VBox(card_children)
    card.add_class('quiz-card')

    def on_check(b):
        is_correct, message = grade_numerical(qid, input_widget.value)
        _update_feedback(feedback_html, is_correct, message, card)

    check_button.on_click(on_check)
    _WIDGET_REGISTRY[qid] = {'card': card, 'input': input_widget, 'feedback': feedback_html}
    return card

def render_mc_question(qid, question_text, options_dict, hint_text=""):
    """Render a multiple-choice question as a widget card."""
    question_html = widgets.HTML(value='<div class="quiz-question-text">' + question_text + '</div>')
    options_list = [(v, k) for k, v in options_dict.items()]  # (description, key)
    input_widget = widgets.RadioButtons(options=options_list, description='', layout={'width': '600px'})
    check_button = widgets.Button(description='✓ Check Answer', button_style='info', layout=widgets.Layout(min_width='180px', height='40px'))
    feedback_html = _create_feedback_html()

    input_row = widgets.HBox([input_widget, check_button])

    card_children = [question_html, input_row, feedback_html]
    if hint_text:
        card_children.append(_create_hint_accordion(hint_text))

    card = widgets.VBox(card_children)
    card.add_class('quiz-card')

    def on_check(b):
        student_choice = input_widget.value
        is_correct, message = grade_mc(qid, student_choice)
        _update_feedback(feedback_html, is_correct, message, card)

    check_button.on_click(on_check)
    _WIDGET_REGISTRY[qid] = {'card': card, 'input': input_widget, 'feedback': feedback_html}
    return card

def render_match_question(qid, intro_text, questions_list, reasoning_pool_dict, hint_text=""):
    """Render a match-the-reasoning question as a widget card."""
    intro_html = widgets.HTML(value='<div class="quiz-question-text">' + intro_text + '</div>')

    # Create a dropdown for each question
    dropdowns = {}
    dropdown_widgets = []
    for q in questions_list:
        qid_sub = q['id']
        q_text = q['text']
        options = [(v + ' [' + k + ']', k) for k, v in reasoning_pool_dict.items()]
        dropdown = widgets.Dropdown(options=options, description=qid_sub + ':')
        dropdowns[qid_sub] = dropdown
        dropdown_widgets.append(dropdown)

    questions_html = widgets.HTML(value='<div style="margin: 8px 0; color: #F4F4F8; line-height: 1.6; font-size: 13px;">' + '<br>'.join([q['id'] + ': ' + q['text'] for q in questions_list]) + '</div>')
    pool_html = widgets.HTML(value='<div style="margin: 8px 0; color: #00E5FF;"><strong>Reasoning Pool:</strong><br>' + '<br>'.join([k + ': ' + v[:80] + '...' if len(v) > 80 else k + ': ' + v for k, v in reasoning_pool_dict.items()]) + '</div>')

    check_button = widgets.Button(description='✓ Check Matches', button_style='info', layout=widgets.Layout(min_width='180px', height='40px'))
    feedback_html = _create_feedback_html()

    card_children = [intro_html, questions_html, pool_html]
    card_children.extend(dropdown_widgets)
    card_children.extend([check_button, feedback_html])
    if hint_text:
        card_children.append(_create_hint_accordion(hint_text))

    card = widgets.VBox(card_children)
    card.add_class('quiz-card')

    def on_check(b):
        student_mapping = {sub_q: dd.value for sub_q, dd in dropdowns.items()}
        n_correct, n_total, message = grade_match(qid, student_mapping)
        is_correct = (n_correct == n_total)
        _update_feedback(feedback_html, is_correct, message, card)

    check_button.on_click(on_check)
    _WIDGET_REGISTRY[qid] = {'card': card, 'input': dropdowns, 'feedback': feedback_html}
    return card

def render_order_question(qid, problem_text, shuffled_steps, hint_text=""):
    """Render an order-the-steps question as a widget card."""
    problem_html = widgets.HTML(value='<div class="quiz-question-text">' + problem_text + '</div>')

    steps_html = widgets.HTML(value='<div style="margin: 8px 0; color: #F4F4F8; line-height: 1.6; font-size: 13px;">' + '<br>'.join(shuffled_steps) + '</div>')

    # Create a dropdown for each position
    step_letters = [s.split(']')[0][1:] for s in shuffled_steps]  # Extract letters like "A", "B", etc.
    n_positions = len(shuffled_steps)
    position_dropdowns = []
    for i in range(n_positions):
        dd = widgets.Dropdown(options=step_letters, description='Position ' + str(i+1) + ':')
        position_dropdowns.append(dd)

    check_button = widgets.Button(description='✓ Check Order', button_style='info', layout=widgets.Layout(min_width='180px', height='40px'))
    feedback_html = _create_feedback_html()

    card_children = [problem_html, steps_html]
    card_children.extend([widgets.HTML(value='<div style="margin: 8px 0; color: #FFD60A; line-height: 1.6; font-size: 13px;">Enter the step letters in the correct order:</div>')])
    card_children.extend(position_dropdowns)
    card_children.extend([check_button, feedback_html])
    if hint_text:
        card_children.append(_create_hint_accordion(hint_text))

    card = widgets.VBox(card_children)
    card.add_class('quiz-card')

    def on_check(b):
        student_order = [dd.value for dd in position_dropdowns]
        n_correct, n_total, message = grade_order(qid, student_order)
        is_correct = (n_correct == n_total)
        _update_feedback(feedback_html, is_correct, message, card)

    check_button.on_click(on_check)
    _WIDGET_REGISTRY[qid] = {'card': card, 'input': position_dropdowns, 'feedback': feedback_html}
    return card

# === v4 NEW: Submit button + analytics dashboard ===

def submit_quiz_v4():
    """Submit the quiz and freeze answers."""
    global _QUIZ_SUBMITTED, _QUIZ_END_TIME
    if _QUIZ_SUBMITTED:
        return "Quiz already submitted."
    _QUIZ_SUBMITTED = True
    _QUIZ_END_TIME = time.time()
    return "Quiz submitted at " + time.strftime("%Y-%m-%d %H:%M:%S")

def render_submit_button():
    """Render the final submit button."""
    submit_button = widgets.Button(description='🎓 SUBMIT QUIZ', button_style='success', layout=widgets.Layout(width='280px', height='56px'))
    submit_button.style = widgets.ButtonStyle(font_size='16px', font_weight='700', text_color='#0B0B1F')
    submit_button.add_class('quiz-submit-button')
    submit_feedback = widgets.HTML(value='')

    def on_submit(b):
        result = submit_quiz_v4()
        submit_feedback.value = '<div class="quiz-feedback-correct">' + result + '</div>'
        # Disable all check buttons
        for qid, reg in _WIDGET_REGISTRY.items():
            if hasattr(reg.get('input', None), 'disabled'):
                pass  # Some inputs are dicts
        # Trigger analytics
        render_analytics_dashboard()

    submit_button.on_click(on_submit)
    return widgets.VBox([submit_button, submit_feedback])

def render_analytics_dashboard():
    """Render the analytics + personalized feedback dashboard."""
    if not _QUIZ_SUBMITTED:
        display(HTML('<div class="quiz-feedback-incorrect">⚠ Please click SUBMIT QUIZ first.</div>'))
        return

    total_time_s = _QUIZ_END_TIME - _QUIZ_START_TIME
    total_time_min = total_time_s / 60

    # Compute analytics (same as v3)
    question_attempts = {}
    for entry in _ATTEMPT_LOG:
        qid = entry["qid"]
        if qid not in question_attempts:
            question_attempts[qid] = {"attempts": 0, "first_correct": None, "final_correct": False, "qtype": entry["qtype"]}
        question_attempts[qid]["attempts"] += 1
        if entry["was_correct"] and question_attempts[qid]["first_correct"] is None:
            question_attempts[qid]["first_correct"] = entry["attempt_num"]
        if entry["was_correct"]:
            question_attempts[qid]["final_correct"] = True

    n_attempted = len(question_attempts)
    n_total_q = len(_NUM_ANSWERS) + len(_MC_ANSWERS) + sum(len(json.loads(base64.b64decode(v).decode())) for v in _MATCH_ANSWERS.values()) + sum(len(json.loads(base64.b64decode(v).decode())) for v in _ORDER_ANSWERS.values())
    n_got_right = sum(1 for q in question_attempts.values() if q["final_correct"])
    n_first_try = sum(1 for q in question_attempts.values() if q["first_correct"] == 1)
    n_needed_retry = sum(1 for q in question_attempts.values() if q["first_correct"] and q["first_correct"] > 1)
    n_never_got = sum(1 for q in question_attempts.values() if not q["final_correct"])
    n_skipped = n_total_q - n_attempted

    # Gap analysis
    gaps_per_q = {}
    for entry in _ATTEMPT_LOG:
        qid = entry["qid"]
        if qid not in gaps_per_q:
            gaps_per_q[qid] = []
        if entry["gap_since_last"] < 100000:
            gaps_per_q[qid].append(entry["gap_since_last"])

    guessing_count = 0
    quick_retry_count = 0
    logical_retry_count = 0
    for qid, gaps in gaps_per_q.items():
        for g in gaps:
            if g < 5:
                guessing_count += 1
            elif g < 60:
                quick_retry_count += 1
            else:
                logical_retry_count += 1

    # Type stats
    type_stats = {}
    for entry in _ATTEMPT_LOG:
        qtype = entry["qtype"]
        if qtype not in type_stats:
            type_stats[qtype] = {"total_attempts": 0, "correct": 0, "wrong": 0, "unique_qs": set()}
        type_stats[qtype]["total_attempts"] += 1
        if entry["was_correct"]:
            type_stats[qtype]["correct"] += 1
        else:
            type_stats[qtype]["wrong"] += 1
        type_stats[qtype]["unique_qs"].add(entry["qid"])

    # Final score
    total_score = 0
    max_score = 0
    for qid in _NUM_ANSWERS:
        max_score += 1
        if qid in question_attempts and question_attempts[qid]["final_correct"]:
            total_score += 1
    for qid in _MC_ANSWERS:
        max_score += 1
        if qid in question_attempts and question_attempts[qid]["final_correct"]:
            total_score += 1
    for qid, encoded in _MATCH_ANSWERS.items():
        correct_mapping = json.loads(base64.b64decode(encoded).decode())
        max_score += len(correct_mapping)
        latest = None
        for entry in _ATTEMPT_LOG:
            if entry["qid"] == qid:
                latest = entry
        if latest and "partial_score" in latest:
            total_score += latest["partial_score"]
    for qid, encoded in _ORDER_ANSWERS.items():
        correct_order = json.loads(base64.b64decode(encoded).decode())
        max_score += len(correct_order)
        latest = None
        for entry in _ATTEMPT_LOG:
            if entry["qid"] == qid:
                latest = entry
        if latest and "partial_score" in latest:
            total_score += latest["partial_score"]
    percent = (total_score / max_score * 100) if max_score > 0 else 0

    # Build HTML dashboard
    html = '<div class="quiz-analytics">'
    html += '<h2>📊 QUIZ ANALYTICS REPORT</h2>'
    html += '<h3>⏱️ Total Time Taken</h3>'
    html += 'Start: ' + time.strftime("%H:%M:%S", time.localtime(_QUIZ_START_TIME)) + '<br>'
    html += 'End:   ' + time.strftime("%H:%M:%S", time.localtime(_QUIZ_END_TIME)) + '<br>'
    html += 'Total: <span class="highlight">' + str(int(total_time_min)) + ' min ' + str(int(total_time_s % 60)) + ' s</span><br><br>'

    html += '<h3>📝 Per-Question Attempt Counts</h3>'
    html += 'Questions attempted: ' + str(n_attempted) + '/' + str(n_total_q) + '<br>'
    html += '<span class="correct">Got right (eventually): ' + str(n_got_right) + '</span><br>'
    html += '<span class="correct">Got right on FIRST try: ' + str(n_first_try) + '</span><br>'
    html += 'Needed retries: ' + str(n_needed_retry) + '<br>'
    html += '<span class="incorrect">Never got right: ' + str(n_never_got) + '</span><br>'
    html += 'Skipped entirely: ' + str(n_skipped) + '<br><br>'

    html += '<h3>🔍 Gap Analysis (guessing vs logical solving)</h3>'
    html += 'Gaps &lt;5s (likely guessing): <span class="incorrect">' + str(guessing_count) + '</span><br>'
    html += 'Gaps 5-60s (quick retry): ' + str(quick_retry_count) + '<br>'
    html += 'Gaps &gt;60s (logical re-solving): <span class="correct">' + str(logical_retry_count) + '</span><br>'
    if guessing_count > 3:
        html += '<br><span class="incorrect">⚠ You seem to be guessing. Slow down and use the hints!</span><br>'
    elif logical_retry_count > 0:
        html += '<br><span class="correct">✓ Good — you are taking time to re-solve problems logically.</span><br>'
    html += '<br>'

    html += '<h3>🎯 Question-Type Repeat Analysis</h3>'
    for qtype in sorted(type_stats.keys()):
        stats = type_stats[qtype]
        type_name = _TYPE_MAP.get(qtype, qtype)
        n_unique = len(stats["unique_qs"])
        avg_attempts = stats["total_attempts"] / n_unique if n_unique > 0 else 0
        html += 'Part ' + qtype + ' (' + type_name + '): ' + str(n_unique) + ' Qs, ' + str(stats["total_attempts"]) + ' attempts, avg ' + str(round(avg_attempts, 1)) + '/Q, <span class="correct">' + str(stats["correct"]) + ' right</span> / <span class="incorrect">' + str(stats["wrong"]) + ' wrong</span><br>'
    html += '<br>'

    html += '<h3>📅 Per-Question Timestamps</h3>'
    for qid in sorted(question_attempts.keys()):
        q_info = question_attempts[qid]
        first_attempt_time = None
        last_attempt_time = None
        for entry in _ATTEMPT_LOG:
            if entry["qid"] == qid:
                if first_attempt_time is None:
                    first_attempt_time = entry["timestamp"]
                last_attempt_time = entry["timestamp"]
        if first_attempt_time and last_attempt_time:
            time_spent = last_attempt_time - first_attempt_time
            status = "✓" if q_info["final_correct"] else "✗"
            status_class = "correct" if q_info["final_correct"] else "incorrect"
            html += '<span class="' + status_class + '">' + status + ' ' + qid + '</span> (' + str(q_info["attempts"]) + ' attempts, ' + str(int(time_spent)) + 's span, first@ ' + time.strftime("%H:%M:%S", time.localtime(first_attempt_time)) + ')<br>'
    html += '<br>'

    html += '<h3>📊 Attempts Per Time Window</h3>'
    time_windows = [(20, "0-20 min"), (40, "20-40 min"), (60, "40-60 min"), (80, "60-80 min"), (100, "80-100 min"), (120, "100-120 min"), (999999, "120+ min")]
    window_counts = [0] * len(time_windows)
    for entry in _ATTEMPT_LOG:
        elapsed_min = (entry["timestamp"] - _QUIZ_START_TIME) / 60
        for i, (boundary, _) in enumerate(time_windows):
            if elapsed_min < boundary:
                window_counts[i] += 1
                break
    for i, (_, label) in enumerate(time_windows):
        if window_counts[i] > 0 or i == 0:
            bar = "█" * min(window_counts[i], 50)
            html += label + ': ' + str(window_counts[i]) + ' ' + bar + '<br>'
    html += '<br>'

    html += '<h2>🏆 FINAL SCORE</h2>'
    score_class = "correct" if percent >= 70 else "highlight"
    html += '<span class="' + score_class + '" style="font-size: 18px;">Score: ' + str(total_score) + '/' + str(max_score) + ' (' + str(round(percent)) + '%)</span><br><br>'

    # Personalized feedback
    html += '<h2>💬 Personalized Feedback</h2>'

    weak_types = []
    strong_types = []
    for qtype in type_stats:
        stats = type_stats[qtype]
        if stats["total_attempts"] == 0:
            continue
        wrong_ratio = stats["wrong"] / stats["total_attempts"]
        avg_attempts = stats["total_attempts"] / len(stats["unique_qs"])
        if wrong_ratio > 0.5 or avg_attempts > 1.5:
            weak_types.append((qtype, wrong_ratio, avg_attempts))
        elif wrong_ratio < 0.2 and avg_attempts < 1.2:
            strong_types.append((qtype, wrong_ratio, avg_attempts))

    if percent >= 90:
        html += '<br><span class="highlight">🏆 CONGRATULATIONS! You excelled (' + str(round(percent)) + '%).</span><br>'
        html += 'You have demonstrated strong mastery.<br><br>'
        html += '<h3>🌟 Advanced Challenge Problems</h3>'
        for i, prob in enumerate(_TOPIC_FEEDBACK.get("advanced_problems", []), 1):
            html += '<br><strong>' + str(i) + '. ' + prob["title"] + '</strong><br>'
            html += prob["problem"] + '<br>'
            html += '<span class="highlight">💡 Hint:</span> ' + prob["hint"] + '<br>'
    elif percent >= 70:
        html += '<br><span class="correct">🎯 Good work (' + str(round(percent)) + '%)!</span> A few targeted improvements will get you to mastery.<br><br>'
    elif percent >= 50:
        html += '<br><span class="highlight">👍 Decent effort (' + str(round(percent)) + '%).</span> You have the basics — let us build on them.<br><br>'
    else:
        html += '<br><span class="incorrect">📚 Physics takes practice (' + str(round(percent)) + '%).</span> Review the slides and try again.<br><br>'

    if weak_types:
        html += '<h3>📍 Areas to Focus On</h3>'
        for qtype, wrong_ratio, avg_attempts in weak_types:
            type_name = _TYPE_MAP.get(qtype, qtype)
            advice = _TOPIC_FEEDBACK.get("type_advice", dict()).get(qtype, dict())
            html += '<br><strong>Part ' + qtype + ' — ' + type_name + '</strong> (' + str(round(wrong_ratio * 100)) + '% wrong, ' + str(round(avg_attempts, 1)) + ' avg attempts)<br>'
            if advice:
                if "common_errors" in advice:
                    html += '<em>⚠ Common errors:</em><br>'
                    for err in advice["common_errors"]:
                        html += '  • ' + err + '<br>'
                if "step_by_step_example" in advice:
                    html += '<em>📝 Step-by-step example:</em><br>'
                    html += '  ' + advice["step_by_step_example"].replace('\n', '<br>  ') + '<br>'
                if "tips" in advice:
                    html += '<em>💡 Tips:</em><br>'
                    for tip in advice["tips"]:
                        html += '  • ' + tip + '<br>'
    html += '<br>'

    if strong_types:
        html += '<h3>💪 Areas of Strength</h3>'
        for qtype, wrong_ratio, avg_attempts in strong_types:
            type_name = _TYPE_MAP.get(qtype, qtype)
            html += 'Part ' + qtype + ' — ' + type_name + ' (' + str(round((1-wrong_ratio) * 100)) + '% correct)<br>'
    html += '<br>'

    mnemonics = _TOPIC_FEEDBACK.get("mnemonics", [])
    if mnemonics:
        html += '<h3>🧠 Mnemonics to Remember</h3>'
        for formula, mnemonic in mnemonics:
            html += '• <strong>' + formula + '</strong> → ' + mnemonic + '<br>'
    html += '<br>'

    html += '<hr>'
    html += '<em>💬 "I tried, and I made it." — William Kamkwamba</em><br>'
    html += 'Keep practicing. Physics builds on itself.<br>'
    html += '</div>'

    display(HTML(html))

# Initialize progress bar and score tracker (displayed at top of dashboard)
global _PROGRESS_BAR, _SCORE_TRACKER
_PROGRESS_BAR = widgets.IntProgress(value=0, description='Progress:', bar_style='info')
_SCORE_TRACKER = widgets.HTML(value='<div style="color: #00E5FF;">📊 Answered: 0 | ✓ Correct: 0</div>')

print("✓ v4 Quiz setup complete. All widgets and grading functions ready.")
print("⏱️ Quiz started at: " + time.strftime("%Y-%m-%d %H:%M:%S"))
print("📊 Total questions: " + str(len(_NUM_ANSWERS) + len(_MC_ANSWERS) + sum(len(json.loads(base64.b64decode(v).decode())) for v in _MATCH_ANSWERS.values()) + sum(len(json.loads(base64.b64decode(v).decode())) for v in _ORDER_ANSWERS.values())))
```
